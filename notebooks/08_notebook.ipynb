# Stepwise Regression

Here I provide a simple stepwise regression function. This procedure selects variables one at a time based on the p-value. The variable with the lowest p-value is selected first. Then the variable with the second-lowest p-value is added. This continues until no remaining variable has a p-value below the p-cutoff.  As a user, you specify the p-cutoff.  The function takes two dataframes (y and X) and a cutoff for the p-value. Any time you wish to do a stepwise regression, you need first to run the following two code cells.

Don't worry about trying to interpret this code.  Scroll down to the application to Hospital Satisfaction to see how the procedure is used.

```python
import pandas
import statsmodels.api as sm
```

```python

def stepwise_regression(y, X, cutoff, features=[], verbose=False):
    for _ in X:
        candidates = set(X.columns) - set(features)
        candidate_pvalues = pandas.Series(index=candidates)

        for candidate in candidates:
            model = sm.OLS(y, sm.add_constant(X[features + [candidate]])).fit()
            candidate_pvalues[candidate] = model.pvalues[candidate]
        best_candidate = candidate_pvalues.idxmin()
        best_pvalue = candidate_pvalues[best_candidate]

        if best_pvalue < cutoff:
            features = features + [best_candidate]
            if verbose:
                print(
                    "Add  {:20} with p-value {:.6f}".format(best_candidate, best_pvalue)
                )
        else:
            break

    return features
```

### Stepwise Example
### Hospital Satisfaction
The following illustrates how to do stepwise regression.  You name your dependent variable y. In this application, the dependent variable is OVERALL. Then you use upper-case X to include all explanatory variables that you want to use in the stepwise procedure. Note that the each explanatory variable is in quotes and the variables are separated by commas. You must also specify the p-value cutoff. The last row of the command specifies that the p-value cutoff be .1.

```python
y = Hospital_Satisfaction.OVERALL
X = Hospital_Satisfaction[
    ["ACCOMODATIONS",  "AGE", "GENDER", "MEALCOURTESY", "MEALSATISFACT", "MEALTASTE","MEALTEMP","MEALTIMING","MEALVARIETY","RECENT",
        "ROOMBEDDING","ROOMCLEANLINESS","ROOMDECOR","ROOMFAMILY","ROOMNOISE_IN","ROOMNOISE_OUT","ROOMPRIVACY","ROOMRESTROOM","ROOMSTAFFCOURTESY",
        "ROOMTEMPERATURE","ROOMTV","WOW"]
]

stepwise_regression(y, X, 0.1, verbose=True)
```

  From immediately above, we see that three variables were selected: ACCOMODATIONS, MEALCOURTESY, AND ROOMBEDDING. With those three variables included, no other explanatory variable has a p-value less than .1. We next run a regression that includes the three selected variables explanatory variables.  Notice that the p-values for the three included variables are different from above. The regression below corrects standard errors for possible presence of heteroskedasticity. The stepwise procedure does not.

```python
RegStepOut=tprstats.model("cs","OVERALL~ACCOMODATIONS+MEALCOURTESY+ROOMBEDDING", Hospital_Satisfaction)
RegStepOut.summary()
```

In some applications, you may wish to require that one or more variables be retained in the model. The command below illustrates how this can be done. The last line in the command requires that ROOMFAMILY and ROOMNOISE_OUT be included. In this application, the stepwise procedure adds the same three variables as were chosen above. This need not always be the case.

```python
y = Hospital_Satisfaction.OVERALL
X = Hospital_Satisfaction[
    ["ACCOMODATIONS",  "AGE", "GENDER", "MEALCOURTESY", "MEALSATISFACT", "MEALTASTE","MEALTEMP","MEALTIMING",
     "MEALVARIETY","RECENT","ROOMBEDDING","ROOMCLEANLINESS","ROOMDECOR","ROOMFAMILY","ROOMNOISE_IN","ROOMNOISE_OUT",
     "ROOMPRIVACY","ROOMRESTROOM","ROOMSTAFFCOURTESY", "ROOMTEMPERATURE","ROOMTV","WOW"]
]

stepwise_regression(y, X, 0.1,features=["ROOMFAMILY","MEALVARIETY"], verbose=True)
```

The regression below has the five variables from the above stepwise analysis.  
The two variables that were required to be included, ROOMFAMILY and MEALVARIETY, have high p-values.  
That is not surprising. If those variables had low p-values, they would have been included without being required.

```python
RegStepOut2=tprstats.model("cs","OVERALL~ROOMFAMILY+MEALVARIETY+ACCOMODATIONS+MEALCOURTESY+ROOMBEDDING", Hospital_Satisfaction)
RegStepOut2.summary()
```

##Natural Gas Demand

As a second illustration of stepwise regression, lets revisit the regression model we studied for the demand for natural gas.  

In the file named Natural_Gas_Data_US_Extended.xlsx, I have retained the variables from Natural_Gas_Data_US.xlsx. In addition, I have included the logarithms of QG, PG, and PCI. I have also included the following sea surface temperatures: R_AMO, R_AMO_1, R_AMO_2, R_ENSO, R_ENSO_1, and R_ENSO_2.  

The command below reads the file and names it NatGasDat.



```python
NatGasDat=pandas.read_excel("data/Natural_Gas_Data_US_Extended.xlsx")
```

The command below requires that the logarithms of price and income be included. The stepwise procedure selects which of the following sea surface temperature variables to include: R_AMO, R_AMO_1, R_AMO_2, R_ENSO, R_ENSO_1, and R_ENSO_2. Recall that AMO denotes Atlantic sea surface temperatures and ENSO denotes Equatorial sea surface temperatures. Also, _1 denotes the value of the variable from the previous year. This is called the one-year lag of the variable. Similarly _2 denotes the value of the variable two years previous. This is called the two-year lag of the variable.  
We use a p-cutoff=.1.

```python
# Perform stepwise regression to aid in determining which ocean surface temperatures to include.
y = NatGasDat.log_QG

X = NatGasDat[
    [ "log_PG", "log_PCI",
    "R_AMO", "R_AMO_1", "R_AMO_2", "R_ENSO", "R_ENSO_1", "R_ENSO_2"
    ]
]
stepwise_regression(y, X, 0.10, features=["log_PG", "log_PCI"], verbose=True)
```

We see that the stepwise procedure has selected R_AMO_1, R_ENSO_1, and R_AMO.

```python
NatGasReg=tprstats.model("ts","log(QG)~log(PG)+log(PCI)+R_AMO+R_AMO_1+R_ENSO_1", NatGasDat)
NatGasReg.summary()
```


Let's investigate including sea surface temperaturs in the Pacific Ocean. Pacific sea surface temperatures oscillate at decadal frequence. Hene, I used PDO to denote Pacific Decadal Oscillation.



```python
#Try also include Pacific Sea Surface temperatures
y = NatGasDat.log_QG
X = NatGasDat[
    [ "log_PG", "log_PCI",
    "R_AMO", "R_AMO_1", "R_AMO_2", "R_ENSO", "R_ENSO_1", "R_ENSO_2", "R_PDO", "R_PDO_1", "R_PDO_2"
    ]
]
stepwise_regression(y, X, 0.10, features=["log_PG", "log_PCI"], verbose=True)
```

The following regression includes the variables selected by the stepwise procedure. We see that all of the variables have very low p-values. Hence, all variables are significant at the 5% level. This regression provides strong evidence that the sea surface temperature variables impact natural gas usage. The natural interpretation is that these variables impact climate and thereby impact natural gas usage.

```python
OurGasReg=tprstats.model("ts","log(QG)~log(PG)+log(PCI)+R_AMO+R_AMO_1+R_ENSO_1", NatGasDat)
OurGasReg.summary()
```
